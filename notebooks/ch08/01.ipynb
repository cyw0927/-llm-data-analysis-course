```python
import matplotlib.pyplot as plt
from matplotlib.patches import (
    PathPatch, Polygon, Ellipse, Circle, Rectangle
)
from matplotlib.path import Path
import matplotlib.transforms as transforms


# =========================================================
# 1. 기본 캔버스
# =========================================================

fig, ax = plt.subplots(figsize=(8, 10))

ax.set_xlim(0, 100)
ax.set_ylim(0, 125)
ax.set_aspect("equal")
ax.axis("off")

# 배경
ax.add_patch(
    Rectangle(
        (0, 0),
        100,
        125,
        facecolor="#f5f1ed",
        edgecolor="none",
        zorder=0
    )
)


# =========================================================
# 2. 색상
# =========================================================

SKIN = "#f2c5a5"
SKIN_SHADOW = "#d99f7c"

HAIR = "#4b1828"
HAIR_LIGHT = "#7f2c3c"
HAIR_DARK = "#241017"

UNIFORM = "#281a22"
UNIFORM_LIGHT = "#38252f"
UNIFORM_SHADOW = "#160f14"

GREEN = "#2da589"
GREEN_DARK = "#19725f"
BLACK = "#151719"

WHITE = "#f7f4ef"
GRAY = "#cbc9c5"
GRAY_DARK = "#777b82"

EYE = "#8e3241"
EYE_DARK = "#34131a"

SCAR = "#853a36"

RED = "#8e2631"
METAL = "#b7bdc4"
METAL_DARK = "#40454b"


# =========================================================
# 3. 공통 함수
# =========================================================

def path_patch(
    verts,
    codes,
    facecolor,
    edgecolor="#181818",
    lw=1.5,
    zorder=1,
    alpha=1.0
):
    """PathPatch 간단 생성 함수"""

    path = Path(verts, codes)

    patch = PathPatch(
        path,
        facecolor=facecolor,
        edgecolor=edgecolor,
        lw=lw,
        zorder=zorder,
        alpha=alpha,
        joinstyle="round",
        capstyle="round"
    )

    ax.add_patch(patch)

    return patch


def closed_curve(points, facecolor, edgecolor="#181818", lw=1.5, zorder=1):
    """
    CURVE4 기반 닫힌 곡선.
    points:
    첫 점 이후 3개 단위로 베지어를 이어가는 식으로 사용.
    """

    codes = [Path.MOVETO]

    for i in range(1, len(points)):
        codes.append(Path.CURVE4)

    # 마지막 닫기
    points = list(points) + [points[0]]
    codes.append(Path.CLOSEPOLY)

    return path_patch(
        points,
        codes,
        facecolor,
        edgecolor,
        lw,
        zorder
    )


# =========================================================
# 4. 체크무늬 넣기
# =========================================================

def add_checker_to_patch(
    clip_patch,
    x0,
    y0,
    x1,
    y1,
    cell=8,
    color1=GREEN,
    color2=BLACK,
    zorder=3
):
    """
    사각 체크무늬를 만든 뒤
    하오리 영역으로 잘라서 표시.
    """

    row = 0
    y = y0

    while y < y1:

        col = 0
        x = x0

        while x < x1:

            color = color1 if (row + col) % 2 == 0 else color2

            rect = Rectangle(
                (x, y),
                cell,
                cell,
                facecolor=color,
                edgecolor="none",
                zorder=zorder
            )

            rect.set_clip_path(clip_patch)

            ax.add_patch(rect)

            x += cell
            col += 1

        y += cell
        row += 1


# =========================================================
# 5. 다리 / 바지
# =========================================================

def draw_legs():

    # -----------------------------
    # 왼쪽 바지
    # -----------------------------

    verts = [
        (43, 57),

        (39, 52),
        (35, 44),
        (33, 35),

        (31, 28),
        (30, 22),
        (33, 18),

        (37, 16),
        (43, 18),
        (46, 25),

        (49, 35),
        (50, 46),
        (49, 57),

        (43, 57)
    ]

    codes = [
        Path.MOVETO,

        Path.CURVE4,
        Path.CURVE4,
        Path.CURVE4,

        Path.CURVE4,
        Path.CURVE4,
        Path.CURVE4,

        Path.CURVE4,
        Path.CURVE4,
        Path.CURVE4,

        Path.CURVE4,
        Path.CURVE4,
        Path.CURVE4,

        Path.CLOSEPOLY
    ]

    path_patch(
        verts,
        codes,
        UNIFORM,
        HAIR_DARK,
        1.7,
        4
    )

    # 왼쪽 주름
    ax.add_patch(
        Polygon(
            [
                (39, 53),
                (44, 47),
                (40, 36),
                (35, 42)
            ],
            closed=True,
            facecolor=UNIFORM_SHADOW,
            edgecolor="none",
            alpha=.6,
            zorder=5
        )
    )

    # -----------------------------
    # 오른쪽 바지
    # -----------------------------

    verts = [
        (49, 57),

        (55, 56),
        (62, 51),
        (66, 43),

        (69, 35),
        (68, 26),
        (65, 20),

        (61, 17),
        (56, 18),
        (53, 25),

        (49, 36),
        (47, 48),
        (49, 57),

        (49, 57)
    ]

    path_patch(
        verts,
        codes,
        UNIFORM,
        HAIR_DARK,
        1.7,
        4
    )

    ax.add_patch(
        Polygon(
            [
                (55, 53),
                (61, 48),
                (61, 38),
                (55, 42)
            ],
            closed=True,
            facecolor=UNIFORM_SHADOW,
            edgecolor="none",
            alpha=.55,
            zorder=5
        )
    )


# =========================================================
# 6. 종아리 붕대
# =========================================================

def draw_leg_wrap(x, y, width, height, angle=0):

    rect = Rectangle(
        (x - width / 2, y),
        width,
        height,
        facecolor=WHITE,
        edgecolor="#bcb7b2",
        lw=1.2,
        zorder=6
    )

    t = transforms.Affine2D().rotate_deg_around(
        x,
        y + height / 2,
        angle
    ) + ax.transData

    rect.set_transform(t)

    ax.add_patch(rect)

    # 붕대 줄
    for i in range(1, 7):

        line = plt.Line2D(
            [
                x - width / 2 + 0.5,
                x + width / 2 - 0.5
            ],
            [
                y + height * i / 7,
                y + height * i / 7
            ],
            color="#d3cec8",
            lw=.8,
            zorder=7
        )

        line.set_transform(t)

        ax.add_line(line)


# =========================================================
# 7. 신발
# =========================================================

def draw_foot(x, y, mirror=False):

    if not mirror:

        points = [
            (x - 6, y),
            (x + 5, y),
            (x + 7, y + 2),
            (x + 5, y + 5),
            (x, y + 6),
            (x - 5, y + 5)
        ]

    else:

        points = [
            (x - 5, y),
            (x + 7, y),
            (x + 8, y + 2),
            (x + 6, y + 5),
            (x, y + 6),
            (x - 5, y + 5)
        ]

    ax.add_patch(
        Polygon(
            points,
            closed=True,
            facecolor="#24191d",
            edgecolor="#120b0d",
            lw=1.3,
            zorder=8
        )
    )

    # 밑창
    ax.plot(
        [x - 5, x + 6],
        [y, y],
        color="#7c7772",
        lw=2.5,
        zorder=9
    )

    # 붉은 끈
    ax.plot(
        [x - 2.5, x, x + 3],
        [y + 4.5, y + 5.5, y + 3],
        color=RED,
        lw=2.5,
        zorder=9
    )


# =========================================================
# 8. 상체
# =========================================================

def draw_torso():

    verts = [
        (41, 79),

        (44, 82),
        (48, 83),
        (52, 82),

        (58, 81),
        (62, 75),
        (61, 66),

        (61, 60),
        (56, 57),
        (49, 57),

        (42, 57),
        (38, 62),
        (39, 71),

        (39, 75),
        (39, 77),
        (41, 79),

        (41, 79)
    ]

    codes = [
        Path.MOVETO,

        Path.CURVE4,
        Path.CURVE4,
        Path.CURVE4,

        Path.CURVE4,
        Path.CURVE4,
        Path.CURVE4,

        Path.CURVE4,
        Path.CURVE4,
        Path.CURVE4,

        Path.CURVE4,
        Path.CURVE4,
        Path.CURVE4,

        Path.CURVE4,
        Path.CURVE4,
        Path.CURVE4,

        Path.CLOSEPOLY
    ]

    path_patch(
        verts,
        codes,
        UNIFORM,
        HAIR_DARK,
        1.6,
        6
    )

    # 옷 중심선
    ax.plot(
        [50, 50],
        [59, 79],
        color="#49333c",
        lw=1,
        zorder=7
    )

    # 단추
    for y in [76, 70, 64]:

        ax.add_patch(
            Circle(
                (50, y),
                0.8,
                facecolor="#b9b9b9",
                edgecolor="#555",
                lw=.6,
                zorder=8
            )
        )

    # 목깃
    ax.add_patch(
        Rectangle(
            (47.5, 79),
            5,
            5,
            facecolor=UNIFORM_SHADOW,
            edgecolor=HAIR_DARK,
            lw=1,
            zorder=8
        )
    )

    ax.plot(
        [48, 52],
        [82.5, 82.5],
        color="#bfbfc2",
        lw=1,
        zorder=9
    )


# =========================================================
# 9. 하오리
# =========================================================

def draw_haori():

    # -----------------------------
    # 왼쪽
    # -----------------------------

    left_verts = [
        (43, 79),

        (34, 80),
        (24, 75),
        (17, 66),

        (10, 56),
        (11, 48),
        (19, 44),

        (27, 40),
        (35, 44),
        (42, 53),

        (45, 61),
        (45, 71),
        (43, 79),

        (43, 79)
    ]

    left_codes = [
        Path.MOVETO,

        Path.CURVE4,
        Path.CURVE4,
        Path.CURVE4,

        Path.CURVE4,
        Path.CURVE4,
        Path.CURVE4,

        Path.CURVE4,
        Path.CURVE4,
        Path.CURVE4,

        Path.CURVE4,
        Path.CURVE4,
        Path.CURVE4,

        Path.CLOSEPOLY
    ]

    left_patch = path_patch(
        left_verts,
        left_codes,
        GREEN,
        "#101010",
        1.7,
        3
    )

    add_checker_to_patch(
        left_patch,
        5,
        35,
        50,
        85,
        8,
        GREEN,
        BLACK,
        3.2
    )

    # -----------------------------
    # 오른쪽
    # -----------------------------

    right_verts = [
        (55, 80),

        (66, 82),
        (78, 77),
        (87, 70),

        (94, 63),
        (91, 54),
        (82, 50),

        (75, 46),
        (67, 47),
        (59, 54),

        (55, 61),
        (53, 71),
        (55, 80),

        (55, 80)
    ]

    right_patch = path_patch(
        right_verts,
        left_codes,
        GREEN,
        "#101010",
        1.7,
        3
    )

    add_checker_to_patch(
        right_patch,
        50,
        40,
        100,
        85,
        8,
        GREEN,
        BLACK,
        3.2
    )

    # 하오리 안쪽
    ax.add_patch(
        Polygon(
            [
                (26, 43),
                (36, 47),
                (41, 53),
                (37, 49),
                (31, 45)
            ],
            closed=True,
            facecolor="#ddd8d2",
            edgecolor="none",
            alpha=.9,
            zorder=4
        )
    )

    ax.add_patch(
        Polygon(
            [
                (65, 48),
                (77, 47),
                (83, 51),
                (72, 50)
            ],
            closed=True,
            facecolor="#ddd8d2",
            edgecolor="none",
            alpha=.9,
            zorder=4
        )
    )


# =========================================================
# 10. 얼굴
# =========================================================

def draw_face():

    # 얼굴 윤곽
    verts = [
        (42, 97),

        (41, 92),
        (42, 86),
        (46, 83),

        (49, 81),
        (54, 82),
        (57, 85),

        (60, 88),
        (60, 94),
        (58, 99),

        (56, 104),
        (51, 106),
        (47, 104),

        (44, 103),
        (42, 101),
        (42, 97),

        (42, 97)
    ]

    codes = [
        Path.MOVETO,

        Path.CURVE4,
        Path.CURVE4,
        Path.CURVE4,

        Path.CURVE4,
        Path.CURVE4,
        Path.CURVE4,

        Path.CURVE4,
        Path.CURVE4,
        Path.CURVE4,

        Path.CURVE4,
        Path.CURVE4,
        Path.CURVE4,

        Path.CURVE4,
        Path.CURVE4,
        Path.CURVE4,

        Path.CLOSEPOLY
    ]

    path_patch(
        verts,
        codes,
        SKIN,
        "#3b2823",
        1.4,
        10
    )

    # 귀
    ax.add_patch(
        Ellipse(
            (41.5, 94),
            3,
            6,
            facecolor=SKIN,
            edgecolor="#4a3029",
            lw=1,
            zorder=9
        )
    )

    ax.add_patch(
        Ellipse(
            (59, 94),
            3,
            6,
            facecolor=SKIN,
            edgecolor="#4a3029",
            lw=1,
            zorder=9
        )
    )

    # 눈 흰자
    ax.add_patch(
        Ellipse(
            (47, 95),
            5.5,
            4.5,
            angle=-6,
            facecolor=WHITE,
            edgecolor=HAIR_DARK,
            lw=1.3,
            zorder=12
        )
    )

    ax.add_patch(
        Ellipse(
            (54, 95),
            5.5,
            4.5,
            angle=6,
            facecolor=WHITE,
            edgecolor=HAIR_DARK,
            lw=1.3,
            zorder=12
        )
    )

    # 홍채
    for x in [47, 54]:

        ax.add_patch(
            Circle(
                (x, 94.8),
                1.65,
                facecolor=EYE,
                edgecolor=EYE_DARK,
                lw=.8,
                zorder=13
            )
        )

        ax.add_patch(
            Circle(
                (x, 94.7),
                .75,
                facecolor="#170b0e",
                edgecolor="none",
                zorder=14
            )
        )

        ax.add_patch(
            Circle(
                (x - .45, 95.4),
                .35,
                facecolor="white",
                edgecolor="none",
                zorder=15
            )
        )

    # 눈썹
    ax.plot(
        [44.7, 47.7],
        [99, 98.2],
        color=HAIR_DARK,
        lw=1.8,
        solid_capstyle="round",
        zorder=14
    )

    ax.plot(
        [52.5, 55.4],
        [98.2, 99],
        color=HAIR_DARK,
        lw=1.8,
        solid_capstyle="round",
        zorder=14
    )

    # 코
    ax.plot(
        [50.7, 50.1, 50.8],
        [94, 91.3, 91],
        color="#b97963",
        lw=.9,
        zorder=14
    )

    # 입
    verts = [
        (47.4, 88.6),
        (49, 87.6),
        (52, 87.5),
        (54, 88.8)
    ]

    mouth_codes = [
        Path.MOVETO,
        Path.CURVE4,
        Path.CURVE4,
        Path.CURVE4
    ]

    path_patch(
        verts,
        mouth_codes,
        "none",
        "#6d3332",
        1.2,
        14
    )

    # 작은 미소
    ax.plot(
        [48, 50, 52.5],
        [88.4, 87.9, 88.3],
        color="#9e5a50",
        lw=.6,
        zorder=13
    )

    # 이마 흉터
    scar = [
        (49, 104),
        (50, 106),
        (52, 106),
        (53.8, 104.6),
        (54, 102.7),
        (52.5, 101.8),
        (50.5, 102.3),
        (49, 104)
    ]

    ax.add_patch(
        Polygon(
            scar,
            closed=True,
            facecolor=SCAR,
            edgecolor="#632521",
            lw=.7,
            zorder=13
        )
    )


# =========================================================
# 11. 머리
# =========================================================

def draw_hair():

    hair_points = [
        (41, 99),
        (38, 103),
        (40, 106),
        (37, 109),

        (43, 109),
        (42, 114),
        (47, 111),

        (49, 117),
        (52, 112),

        (57, 116),
        (57, 111),

        (63, 113),
        (60, 108),

        (64, 108),
        (60, 103),

        (60, 99),
        (57, 102),

        (55, 106),
        (52, 107),
        (48, 106),

        (45, 103),
        (41, 99)
    ]

    ax.add_patch(
        Polygon(
            hair_points,
            closed=True,
            facecolor=HAIR,
            edgecolor=HAIR_DARK,
            lw=1.8,
            joinstyle="round",
            zorder=16
        )
    )

    # 앞머리
    bangs = [
        [
            (42.5, 103),
            (46, 99),
            (46, 106)
        ],

        [
            (46, 105),
            (49, 100),
            (50, 107)
        ],

        [
            (49, 106),
            (52, 100),
            (53.5, 106)
        ],

        [
            (53, 106),
            (56, 102),
            (57, 106)
        ]
    ]

    for b in bangs:

        ax.add_patch(
            Polygon(
                b,
                closed=True,
                facecolor=HAIR,
                edgecolor=HAIR_DARK,
                lw=1,
                zorder=17
            )
        )

    # 머리 하이라이트
    ax.plot(
        [43, 47, 50],
        [108, 112, 111],
        color=HAIR_LIGHT,
        lw=2,
        alpha=.7,
        zorder=18
    )

    ax.plot(
        [53, 57, 59],
        [111, 113, 109],
        color=HAIR_LIGHT,
        lw=1.7,
        alpha=.65,
        zorder=18
    )


# =========================================================
# 12. 귀걸이
# =========================================================

def draw_earring(x, y, mirror=False):

    ax.plot(
        [x, x],
        [y, y - 5],
        color="#33252a",
        lw=1,
        zorder=15
    )

    width = 4

    earring = Polygon(
        [
            (x - width / 2, y - 5),
            (x + width / 2, y - 5),
            (x + width / 2 + .5, y - 12),
            (x - width / 2 - .5, y - 12)
        ],
        closed=True,
        facecolor=WHITE,
        edgecolor="#4b4e53",
        lw=.9,
        zorder=14
    )

    ax.add_patch(earring)

    # 붉은 태양
    ax.add_patch(
        Circle(
            (x, y - 7),
            .65,
            facecolor="#b82931",
            edgecolor="none",
            zorder=15
        )
    )

    # 문양
    ax.plot(
        [x - 1.8, x + 1.8],
        [y - 9.3, y - 11],
        color="#7f8388",
        lw=.7,
        zorder=15
    )


# =========================================================
# 13. 팔 + 손
# =========================================================

def draw_arms():

    # 왼팔
    ax.add_patch(
        Polygon(
            [
                (42, 76),
                (46, 76),
                (53, 68),
                (50, 64),
                (43, 67),
                (39, 71)
            ],
            closed=True,
            facecolor=UNIFORM,
            edgecolor=HAIR_DARK,
            lw=1.3,
            zorder=7
        )
    )

    # 오른팔
    ax.add_patch(
        Polygon(
            [
                (56, 77),
                (61, 73),
                (63, 68),
                (59, 64),
                (54, 68),
                (51, 72)
            ],
            closed=True,
            facecolor=UNIFORM,
            edgecolor=HAIR_DARK,
            lw=1.3,
            zorder=7
        )
    )

    # 손
    ax.add_patch(
        Ellipse(
            (51.3, 66),
            5,
            7,
            angle=30,
            facecolor=SKIN,
            edgecolor="#6f493c",
            lw=1,
            zorder=12
        )
    )

    ax.add_patch(
        Ellipse(
            (57.2, 67.7),
            5,
            7,
            angle=-25,
            facecolor=SKIN,
            edgecolor="#6f493c",
            lw=1,
            zorder=12
        )
    )


# =========================================================
# 14. 벨트
# =========================================================

def draw_belt():

    ax.add_patch(
        Rectangle(
            (41, 55),
            20,
            4,
            facecolor="#d7d5d2",
            edgecolor="#777",
            lw=1,
            zorder=9
        )
    )

    # 버클
    ax.add_patch(
        Rectangle(
            (47, 55.3),
            6,
            3.4,
            facecolor="#eeeeec",
            edgecolor="#777",
            lw=1,
            zorder=10
        )
    )


# =========================================================
# 15. 검
# =========================================================

def draw_sword():

    # 손잡이
    handle = Polygon(
        [
            (43, 61),
            (45, 58.5),
            (56, 66),
            (54, 68.5)
        ],
        closed=True,
        facecolor="#25292d",
        edgecolor="#101113",
        lw=1.2,
        zorder=14
    )

    ax.add_patch(handle)

    # 손잡이 빨간 끈
    for i in range(5):

        x = 46 + i * 2

        ax.plot(
            [x, x + 1.6],
            [60.5 + i * 1.4, 62.8 + i * 1.4],
            color="#a32934",
            lw=1.5,
            zorder=15
        )

    # 츠바
    ax.add_patch(
        Ellipse(
            (56.5, 67.5),
            6,
            4.5,
            angle=40,
            facecolor="#41271d",
            edgecolor="#17100c",
            lw=1.2,
            zorder=15
        )
    )

    ax.add_patch(
        Circle(
            (56.5, 67.5),
            1.3,
            facecolor="#9c4c28",
            edgecolor="#54220f",
            lw=.8,
            zorder=16
        )
    )

    # 검날
    blade = Polygon(
        [
            (57.5, 69),
            (59.2, 70.1),
            (94, 110),
            (92.5, 112)
        ],
        closed=True,
        facecolor=METAL,
        edgecolor=METAL_DARK,
        lw=1.3,
        zorder=13
    )

    ax.add_patch(blade)

    # 검날 어두운 면
    ax.add_patch(
        Polygon(
            [
                (58.6, 70),
                (60, 70.8),
                (94, 110),
                (92.7, 109)
            ],
            closed=True,
            facecolor="#3c4148",
            edgecolor="none",
            zorder=14
        )
    )

    # 날 빛
    ax.plot(
        [60, 93],
        [71.5, 110],
        color="#e3e6e8",
        lw=.8,
        alpha=.8,
        zorder=15
    )


# =========================================================
# 16. 전체 그리기
# =========================================================

# 뒤쪽 하오리
draw_haori()

# 다리
draw_legs()

# 붕대
draw_leg_wrap(37, 10, 9, 12, angle=-8)
draw_leg_wrap(61, 11, 8, 11, angle=5)

# 신발
draw_foot(36, 5)
draw_foot(62, 7, mirror=True)

# 상체
draw_torso()

# 팔
draw_arms()

# 허리
draw_belt()

# 얼굴
draw_face()

# 귀걸이 - 머리보다 일부는 뒤
draw_earring(41.3, 95)
draw_earring(59.3, 95)

# 머리
draw_hair()

# 검
draw_sword()


# =========================================================
# 17. 작은 그림자
# =========================================================

ax.add_patch(
    Ellipse(
        (49, 4),
        48,
        5,
        facecolor="#888888",
        edgecolor="none",
        alpha=.15,
        zorder=1
    )
)


# =========================================================
# 18. 출력 / 저장
# =========================================================

plt.tight_layout()

plt.savefig(
    "tanjiro_improved.png",
    dpi=250,
    bbox_inches="tight",
    pad_inches=0.05
)

plt.show()
```

```python
from pathlib import Path
import pandas as pd
import numpy as np

DATA_DIR = Path("data/raw")

customers = pd.read_csv(DATA_DIR / "customers.csv")
products = pd.read_csv(DATA_DIR / "products.csv")
orders = pd.read_csv(DATA_DIR / "orders.csv")
order_items = pd.read_csv(DATA_DIR / "order_items.csv")

print(DATA_DIR.resolve())

display(customers.head())
display(products.head())
display(orders.head())
display(order_items.head())
```

```python
order_items = order_items.copy()
order_items["line_total"] = (
    order_items["quantity"]
    * order_items["unit_price"]
)

print(order_items.info())
print(
    (order_items["line_total"] == 
    order_items["quantity"] * order_items["unit_price"]).sum()
    )
```

```python
completed_orders = orders[
    orders["order_status"].eq("completed")
].copy()

completed_orders.head()
```

```python

visualization_df = (
    order_items
    .merge(
        completed_orders,
        on="order_id",
        how="inner",
        validate="many_to_one",
    )

    .merge(
        products[["product_id", "product_name", "category"]],
        on="product_id",
        how="left",
        validate="many_to_one",
    )

    .merge(
        customers[["customer_id", "age", "gender", "city"]],
        on="customer_id",
        how="left",
        validate="many_to_one",
    )

)

visualization_df["order_date"] = pd.to_datetime(
    visualization_df["order_date"],
    errors="coerce",
)

visualization_df["order_month"] = (
    visualization_df["order_date"]
    .dt.to_period("M")
    .astype(str)
)
```

```python
visualization_df.head()
```

```python

monthly_summary = (

    visualization_df
    .groupby("order_month", as_index=False)
    .agg(
        total_amount=("line_total", "sum"),
        order_count=("order_id", "nunique"),
    )
    .sort_values("order_month")

)

monthly_summary.head()
```

```python

category_summary = (
    visualization_df
    .groupby("category", as_index=False)
    .agg(
        total_amount=("line_total", "sum"),
        total_quantity=("quantity", "sum"),
        order_count=("order_id", "nunique"),
    )
    .sort_values("total_amount", ascending=False)
)

category_summary.head()
```

```python

product_summary = (

    visualization_df
    .groupby(["product_id", "product_name"], as_index=False)
    
    .agg(
        total_amount=("line_total", "sum"),
        total_quantity=("quantity", "sum"),
        order_count=("order_id", "nunique"),
    )
    .sort_values("total_amount", ascending=False)
)

product_summary.head()
```

```python
order_summary = (

    visualization_df
    .groupby("order_id", as_index=False)

    .agg(

        order_date=("order_date", "first"),
        payment_method=("payment_method", "first"),
        age=("age", "first"),
        city=("city", "first"),
        order_total=("line_total", "sum"),
        total_quantity=("quantity", "sum"),
    )
)

order_summary.head()
```

```python

payment_summary = (
    
    order_summary
    .groupby("payment_method", as_index=False)
    .agg(order_count=("order_id", "nunique"))
    .sort_values("order_count", ascending=False)

)

payment_summary.head()
```
